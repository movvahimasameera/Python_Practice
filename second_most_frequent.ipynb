```python

def second_most_frequent(nums):

    if not nums:
        return None

    d = {}

    for i in nums:
        d[i] = d.get(i, 0) + 1

    highest = max(d.values())

    second = None

    for count in d.values():
        if count < highest:
            if second is None or count > second:
                second = count

    if second is None:
        return None

    for i in nums:
        if d[i] == second:
            return i

    return None
```

```python
print(second_most_frequent([4, 5, 4, 5, 4, 6, 6, 7]))
```
